# ResearchPlot 1.0 gallery

This output-free notebook demonstrates the target-oriented workflow. Run it from the repository root after installing `researchplot-venues`.

```python
import matplotlib.pyplot as plt

import researchplot as rp
```

```python
target = rp.target(
    "ieee-journal@2026.08.0",
    role="main",
    width="single",
    content="line-art",
)

with target.style() as style:
    fig, ax = style.subplots(aspect=0.68)
    x = [0, 1, 2, 3, 4]
    ax.plot(x, [value**2 for value in x], marker="o", label="Quadratic")
    ax.plot(x, [2 * value + 1 for value in x], marker="s", linestyle="--", label="Linear")
    ax.set(xlabel="Input", ylabel="Response")
    ax.legend(frameon=False)

report = target.validate(fig)
print(report.verdict)
fig
```

```python
result = target.export(fig, "build/examples/gallery-figure.pdf", policy="complete")
print(result.paths)
print(result.manifest_path)
plt.close(fig)
```
