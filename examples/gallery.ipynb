# Example with all the plots

```python
# Basic Plots
import numpy as np

from researchplot import plots

# Sample data
values = [10, 20, 30, 40]
categories = ["A", "B", "C", "D"]
x = np.linspace(0, 10, 100)
y = np.sin(x)
matrix = np.random.rand(10, 10)
cm = np.array([[50, 10], [5, 35]])
X, Y = np.meshgrid(np.linspace(-2, 2, 100), np.linspace(-2, 2, 100))
Z = np.sin(X) * np.cos(Y)
pie_values = [0.3, 0.4, 0.3]
pie_categories = ["Category 1", "Category 2", "Category 3"]

plots.bar(values, xticks=categories, format="ieee", title="IEEE Bar Plot")
plots.stacked_bar(
    [[25, 30], [15, 40]],
    stack_labels=["Group 1", "Group 2"],
    format="nature",
    title="Nature Stacked Bar",
)
plots.scatter(
    x,
    y,
    xlabel="X Axis",
    ylabel="Y Axis",
    format="springer",
    show_correlation=True,
    title="Springer Scatter Plot",
)
plots.line(
    x,
    np.sin(x),
    xlabel="Time",
    ylabel="Value",
    format="science",
    show_confidence_interval=True,
    ci_data=[y + np.random.normal(0, 0.1, 100) for _ in range(10)],
    title="Science Line Plot",
)

# Distribution Plots
plots.histogram(
    np.random.normal(0, 1, 1000), bins=20, format="elsevier", title="Elsevier Histogram"
)
plots.boxplot(
    [np.random.normal(0, 1, 100) for _ in range(4)],
    labels=categories,
    format="cell",
    title="Cell Boxplot",
)
plots.violinplot(
    [np.random.normal(0, 1, 100) for _ in range(3)],
    labels=["A", "B", "C"],
    format="pnas",
    title="PNAS Violin Plot",
)

# Matrix/Heatmap Plots
plots.heatmap(matrix, format="ieee", title="IEEE Style Heatmap")
plots.confusion_matrix(
    cm, classes=["Class 0", "Class 1"], format="nature", title="Nature Confusion Matrix"
)

# Time Series/Sequential Data
epochs = np.arange(1, 21)
train_acc = np.linspace(0.5, 0.95, 20)
val_acc = (train_acc - np.random.normal(0, 0.03, 20)).tolist()
plots.accuracy_vs_epoch(
    epochs, train_acc, val_acc, format="springer", title="Springer Accuracy vs Epoch"
)

# Advanced Plots
# 3D Surface
plots.surface_3d(
    X, Y, Z, xlabel="X", ylabel="Y", zlabel="Z", format="science", title="Science 3D Surface"
)

# Sankey Diagram (requires specific flow data)
flows = [0.25, 0.15, 0.60, -0.20, -0.15, -0.25, -0.5]
labels = ["Input", "Process 1", "Process 2", "Loss 1", "Output", "Loss 2", "Storage"]
plots.sankey(flows, labels, format="ieee", title="IEEE Sankey Diagram")

# Error Band Plot
x = np.linspace(0, 10, 100)
y_mean = np.sin(x)
y_std = 0.2 * np.abs(np.cos(x))
plots.error_band(x, y_mean, y_std, format="nature", title="Nature Error Band Plot")

# Radar Chart
plots.radar_chart(categories, values, format="springer", title="Springer Radar Chart")

# Contour Plot
plots.contour_plot(
    X, Y, Z, levels=15, format="science", contour_cmap="viridis", title="Science Contour Plot"
)

# Pie Chart
plots.pie(pie_values, pie_categories, format="elsevier", title="Elsevier Pie Chart")

# Hexbin Plot
plots.hexbin(
    np.random.normal(0, 1, 1000),
    np.random.normal(0, 1, 1000),
    format="cell",
    title="Cell Hexbin Density",
)

# Learning Curves
train_sizes = np.linspace(0.1, 1.0, 10)
train_scores = np.linspace(0.4, 0.95, 10)
test_scores = np.linspace(0.3, 0.85, 10)
plots.learning_curves(
    train_sizes, train_scores, test_scores, format="ieee", title="IEEE Learning Curves"
)

# Quiver Plot
X, Y = np.meshgrid(np.linspace(-2, 2, 100), np.linspace(-2, 2, 100))
U = np.sin(X)
V = np.cos(Y)
plots.quiver(X, Y, U, V, format="nature", title="Nature Quiver Plot")
```
